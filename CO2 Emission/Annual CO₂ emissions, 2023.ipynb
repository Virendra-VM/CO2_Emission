# Annual CO₂ emissions

```python
import pandas as pd
```

```python
import numpy as np
```

```python
import plotly.express as px
```

```python
import plotly.graph_objects as go
```

```python
co2 = pd.read_csv('annual-co2-emissions-per-country.csv')
```

```python
co2
```

```python
selected_countries = ['United States', 'India', 'Germany', 'Brazil', 'United Kingdom', 'France','China']
filtered_co2 = co2[co2['Entity'].isin(selected_countries)]

# Plot the line chart
fig = px.line(
    filtered_co2,
    x='Year', 
    y='Annual CO₂ emissions',  
    color='Entity',
    title='Annual CO2 Emissions for Selected Countries',
    labels={'Year': 'Year', 'CO2 Emissions': 'CO2 Emissions (metric tons)'}
)

fig.show()
```

```python
total_emissions = co2.groupby('Entity', as_index=False)['Annual CO₂ emissions'].sum()

fig = px.choropleth(
    total_emissions,
    locations='Entity',  
    locationmode='country names', 
    color='Annual CO₂ emissions',  
    hover_name='Entity',
    hover_data={'Annual CO₂ emissions': True, 'Entity': False},
    title=f'Total CO2 Emissions by Country',
     color_continuous_scale=[
        "#ffe6c2", "#f5a259", "#cd8142", "#a36228", "#7a4414", "#4e2600"
    ] 
    
)
fig.show()
```

```python
# Ensure the columns for 'Year', 'Country', and 'CO2 Emissions' are correctly named
# If not, replace these with the actual column names
co2 = co2.rename(columns={'Year': 'Year', 'Entity': 'Country', 'Annual CO₂ emissions': 'CO2 Emissions'})

earliest_year = co2['Year'].min()
latest_year = co2['Year'].max()
custom_color_scale = [
    [0.0, "#f8f0e3"],   # Lightest brown (3 million t or less)
    [0.1, "#e3c5a7"],   # 10 million t
    [0.2, "#d0a07c"],   # 30 million t
    [0.3, "#bf8056"],   # 100 million t
    [0.4, "#a15e3e"],   # 300 million t
    [0.6, "#7c4025"],   # 1 billion t
    [0.8, "#532613"],   # 3 billion t
    [1.0, "#301104"],   # Darkest brown (10 billion t or more)
]

# Create the interactive choropleth map with a time slider
fig = px.choropleth(
    co2,
    locations='Country',  # Column with country names
    locationmode='country names',  # Use country names for mapping
    color='CO2 Emissions',  # Emissions to determine color intensity
    #animation_frame='Year',  # Create a time slider based on year
    hover_name='Country',  # Display country name on hover
    hover_data={'CO2 Emissions': True},  # Show CO2 emissions on hover
    title='Cumulative CO2 Emissions Over Time',
    color_continuous_scale=custom_color_scale   # Light to dark shades of red
)

# Add layout with slider for year selection
fig.update_layout(
    sliders=[dict(
        active=0,
        currentvalue={"prefix": "Year: "},
        pad={"t": 50},
        steps=[
            dict(
                label=str(year),
                method="restyle",
                args=[{"z": [co2[co2['Year'] == year]['CO2 Emissions']],
                       "locations": [co2[co2['Year'] == year]['Country']]}]
            )
            for year in range(earliest_year, latest_year + 1)
        ]
    )],
    geo=dict(showframe=False, showcoastlines=False),
    title="Cumulative CO2 Emissions by Country",
    coloraxis_colorbar=dict(
        title="CO2 Emissions (t)",
        ticks="outside"
    )
)

# Show the map
fig.show()
```
